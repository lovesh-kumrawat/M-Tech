```python
class Student:

    # Constructor
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # Method
    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

    def __repr__(self):
        return f"Student(name={self.name}, marks={self.marks})"


# ---------------------------------
# STEP 2: Create Objects
# ---------------------------------
s1 = Student("Rahul", 85)
s2 = Student("Priya", 90)

# ---------------------------------
# STEP 3: Use Object
# ---------------------------------
s1.display()
s2.display()
```

```python
class Student:

    def __init__(self, name):
        self.name = name
        print("Constructor called")

    def __del__(self):
        print("Destructor called for", self.name)


# Create object
s1 = Student("Rahul")

# Delete object
del s1
```

```python
class FileHandler:

    def __init__(self, filename):
        self.file = open(filename, "w")
        print("File opened")

    def write_data(self):
        self.file.write("Hello Students")

    def __del__(self):
        self.file.close()
        print("File closed automatically")


f = FileHandler("demo.txt")
f.write_data()

del f
```
