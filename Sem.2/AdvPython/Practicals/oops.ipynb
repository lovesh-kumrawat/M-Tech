```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
        print("Called __init__")

    def display(self):
        print(f"Name: {self.name}")
        print(f"Marks: {self.marks}")

    def __del__(self):
        print("Called __del__")

s1 = Student("Aditya", 25)
s2 = Student("Akshay", 45)

s1.display()
s2.display()
```

```python
del s1, s2, Student
```

```python
def call():
    s1 = Student("Aditya", 25)
    s2 = Student("Akshay", 45)

    s1.display()
    s2.display()
```

```python
call()
```
